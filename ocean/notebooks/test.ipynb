```python
import numpy as np
import os
import shutil
import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
if os.path.exists('../data/dataset'):
    shutil.rmtree('../data/dataset')

os.makedirs('../data/dataset')
os.makedirs('../data/dataset/hr/psi1')
os.makedirs('../data/dataset/lr/psi1')
os.makedirs('../data/dataset/hr/psi2')
os.makedirs('../data/dataset/lr/psi2')

folder_path = '../data/sr_dataset/hr_320_544'
paths = os.listdir(folder_path)
for i, path in enumerate(paths):
    arr = np.load(folder_path + '/' + path)
    for j, sample in enumerate(arr):
        sample = sample.reshape(544, 320)
        if j < 150:
            # then we are in psi1
            np.save(f'../data/dataset/hr/psi1/{i}_{j}.npy', sample)
        else:
            # then we are in psi2
            np.save(f'../data/dataset/hr/psi2/{i}_{j}.npy', sample)

folder_path = '../data/sr_dataset/lr_40_68'
paths = os.listdir(folder_path)
for i, path in enumerate(paths):
    arr = np.load(folder_path + '/' + path)
    for j, sample in enumerate(arr):
        sample = sample.reshape(68, 40)
        if j < 150:
            # then we are in psi1
            np.save(f'../data/dataset/lr/psi1/{i}_{j}.npy', sample)
        else:
            # then we are in psi2
            np.save(f'../data/dataset/lr/psi2/{i}_{j}.npy', sample)
```
